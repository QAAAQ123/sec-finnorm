```python
def get_filings_by_date(amendments: bool = True,*, companies: list[str],start: str,end: str,):
    return [
        Company(company).get_filings().filter(
            filing_date=f"{start}:{end}",
            form=["10-K","10-Q"])
        for company in companies
    ]
```

```python
filings = get_filings_by_date(
    companies=["AAPL","GOOGL"],
    start="2025-10-18",
    end="2026-10-18"
    )
print(type(filing for filing in filings))
```
