```python
companies = ["AAPL", "NVDA", "O", "1652044"]
companies_entity_list: list[Company] = [Company(c_identifier)for c_identifier in companies]
```

```python
print(companies_entity_list)
```

```python
company_latest_financials = [
    company.get_financials()
    for company in companies_entity_list
]
```

```python
from IPython.display import display

for financials in company_latest_financials:
    display(financials.cash_flow_statement().to_markdown())
```
